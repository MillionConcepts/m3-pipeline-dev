```python
import matplotlib.pyplot as plt

plt.figure(figsize=(9, 6))
chann = 0

# Colorblind-friendly palette (Okabe-Ito scheme + additional colors)
# These colors are distinguishable for people with red-green colorblindness
colors = {
    'l1b': '#E69F00',           # Orange (distinguishable from all others)
    'l0_baseline': '#56B4E9',   # Sky blue
    'lab_flat': '#009E73',      # Teal green
    'obs_flat': '#D55E00',      # Reddish-orange (still distinct in CVD)
    'combined': '#CC79A7',      # Rose/mauve
    'ratio': '#0072B2',         # Dark blue
}

plt.plot(range(304), l1b[1000, chann, :], 
         color=colors['l1b'], label='l1b', linewidth=2)
plt.plot(range(304), l0[1000, chann, :] + .5, 
         color=colors['l0_baseline'], label='l0 to l1b without flats', linewidth=2)
plt.plot(range(304), l0[1000, chann, :] * lab_flat[chann+1, 9:313] + 1, 
         label='image*lab_flat', color=colors['lab_flat'], linewidth=2)
plt.plot(range(304), l0[1000, chann, :] * obs_flat[chann+1, 9:313] + 2, 
         label='image*obs_flat', color=colors['obs_flat'], linewidth=2)
plt.plot(range(304), l0[1000, chann, :] * lab_flat[chann+1, 9:313] * obs_flat[chann+1, 9:313] - 1.5, 
         label='image*obs_flat*lab_flat', color=colors['combined'], linewidth=2)
plt.plot(range(304), l0[1000, chann, :] * lab_flat[chann+1, 9:313] / obs_flat[chann+1, 9:313]+23, 
         color=colors['ratio'], label='image*lab_flat/obs_flat', linewidth=2)

plt.xlabel(f"crosstrack on channel {chann}", fontsize=12)
plt.ylabel("Value", fontsize=12)
plt.legend(loc='best', framealpha=0.5)
plt.grid(True, alpha=0.3)  # Optional: add grid for better readability
plt.tight_layout()
plt.show()
```
