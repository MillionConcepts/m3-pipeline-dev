```python
import numpy as np
import matplotlib.pyplot as plt
from astropy.io import fits
```

```python
path = '/home/bekah/m3-pipeline-dev/data/dark/temp_binned_analysis/global_1k_groups/dark_temp_144.0-145.0K_median.fits'
with fits.open(path) as hdul:
        image0 = hdul[0].data 

path = '/home/bekah/m3-pipeline-dev/data/dark/temp_binned_analysis/m3t/dark_temp_146.0-147.0K_median.fits'
with fits.open(path) as hdul:
        image1 = hdul[0].data 

path = '/home/bekah/m3-pipeline-dev/data/dark/temp_binned_analysis/global_1k_groups/dark_temp_162.0-163.0K_median.fits'
with fits.open(path) as hdul:
        image2 = hdul[0].data 

path = '/home/bekah/m3-pipeline-dev/data/dark/temp_binned_analysis/m3t/dark_temp_162.0-163.0K_median.fits'
with fits.open(path) as hdul:
        image3 = hdul[0].data 

path = '/home/bekah/m3-pipeline-dev/data/dark/temp_binned_analysis/global_1k_groups/dark_temp_156.0-157.0K_median.fits'
with fits.open(path) as hdul:
        image4= hdul[0].data 

path = '/home/bekah/m3-pipeline-dev/data/dark/temp_binned_analysis/m3t/dark_temp_160.0-161.0K_median.fits'
with fits.open(path) as hdul:
        image5 = hdul[0].data 
```

```python

fig, axes = plt.subplots(3, 2, figsize=(10, 8))

images = [image0, image1, image4, image5, image2, image3]
axes_flat = [axes[0, 0], axes[0, 1], axes[1, 0], axes[1, 1], axes[2,0], axes[2,1]]
labels = ['144-145 K', '144-145 K', '156-157 K', '160-161 K', '162-163 K', '162-163 K']
col_labels = ['Global', 'Target']

for ax, img, label in zip(axes_flat, images, labels):
    vmin, vmax = np.percentile(img, [5, 90])
    im = ax.imshow(img, cmap='gray', vmin=vmin, vmax=vmax)
    cbar = fig.colorbar(im, ax=ax, orientation='horizontal', fraction=0.046, pad=0.12)
    cbar.ax.tick_params(labelsize=6)
    ax.tick_params(axis='both', labelsize=6)
    ax.set_title(label, fontsize=12)

for ax, col_label in zip(axes[0], col_labels):
    ax.annotate(col_label, xy=(0.5, 1.3), xycoords='axes fraction',
                ha='center', va='bottom', fontsize=12, fontweight='bold')

#plt.tight_layout()
plt.savefig("different_temps_target_global.png", dpi=400)

plt.show()

```
