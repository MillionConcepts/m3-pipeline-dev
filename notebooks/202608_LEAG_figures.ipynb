```python
import matplotlib.pyplot as plt 
from astropy.io import fits 
import numpy as np 
import pandas as pd 
```

```python
# change to your local dir where downloaded data / images are 

local_dir = "/home/bekah/m3-pipeline-dev/data/leag_pics"
```

```python
#### Bad Dark Signal Images (broken star tracker, dark has unexpected illumination) 

# Dark Signal 
with fits.open(f"{local_dir}/m3g20090601t102652_l0.fits") as hdul:
    # let's only look at one band for this 
    dark_obs = hdul[0].data[63, :, :]

plt.figure(figsize=(8, 9))

# add a stretch to show illuminated area better 
plt.imshow(dark_obs, cmap='gray', vmin=500, vmax=650)
plt.title('Dark Observation M3G20090601T102652 Band 63')
plt.axis('off')

plt.savefig("baddarksignal_band63.png", dpi=300) 
```

```python
#### Bad Dark Signal Images (broken star tracker, dark has unexpected illumination) Flags

# Corresponding BDE / Flag Map 
with fits.open(f"{local_dir}/m3g20090601t102652_bde.fits") as hdul:
    flags = hdul[0].data

# more normal BDE Map 
with fits.open(f"{local_dir}/m3g20090603t215712_bde.fits") as hdul:
    flags_norm = hdul[0].data

fig, (ax2, ax3) = plt.subplots(2, 1, figsize=(8, 4))

im2 = ax2.imshow(flags, cmap='gray')
ax2.set_title('M3G20090601T102652 Bad Detector Element Map')
ax2.axis('off')

im3 = ax3.imshow(flags_norm, cmap='gray')
ax3.set_title('Normal Bad Detector Element Map')
ax3.axis('off')


plt.tight_layout()

# plt.show()

plt.savefig("baddarksignal_bde_maps.png", dpi=300) 
```

```python
##### Dark Signal over time vs temperature 
# median and std 

from mpl_toolkits.axes_grid1.inset_locator import inset_axes

dark_df = pd.read_csv(f"{local_dir}/m3_stats_dark.csv")

# higher bands are unnecessarily messy for this plot (ie at higher bands at high temps sometimes the median is negative
# and that makes it annoying to show the whole plot). 
dark_df = dark_df[dark_df['band'] < 70] 

fig, ax = plt.subplots(figsize=(10, 6))

# only use one band for dark signal obs counts, otherwise counts will be correct count x 86 
ax.hist(dark_df[dark_df['band']==0]['temperature'], bins=20, color='lightgray', edgecolor='white', alpha=0.5)

ax1 = ax.twinx()

# temp vs median center cols per band for dark 
dat = ax1.scatter(
    dark_df["temperature"],
    dark_df["median_cols_20_300"],
    s=4,
    c=dark_df["band"],
    cmap="cividis",
)

# put colorbar inside plot to save space
axins = inset_axes(
    ax1,
    width="50%",          
    height="5%",        
    loc="upper center",   
    borderpad=1.5 
)

plt.colorbar(dat, cax=axins, label="Band", orientation='horizontal')

ax.set_ylabel("Dark Signal Observation Counts") 

ax1.set_xlabel("Observation Temperature (K)")
ax1.set_ylabel("Median Dark Signal DN per Band (Samples 20:300)") #, color='navy')
ax1.set_title("Dark Signal DNs vs Observation Temperature")

plt.show()

fig.savefig("darksignal_vs_temp.png", dpi=300) 
```

```python
#### Cold, Intermediate, and Warm Dark Signal Images 

# Mid temp dark 
with fits.open(f"{local_dir}/m3g20090121t025911_l0.fits") as hdul:
    mid = hdul[0].data.transpose(1,0,2)

# Cold dark 
with fits.open(f"{local_dir}/m3g20090816t045533_l0.fits") as hdul:
    cold = hdul[0].data.transpose(1,0,2)

# Hot dark 
with fits.open(f"{local_dir}/m3g20090616t073912_l0.fits") as hdul:
    hot = hdul[0].data.transpose(1,0,2)

fig, (ax1, ax2, ax3) = plt.subplots(3, 1, figsize=(9, 9))

im1 = ax1.imshow(np.median(cold, axis=0), cmap='gray', vmin=500, vmax=600)
ax1.set_title('M3G20090816T045533, 144.74 K')
ax1.axis('off')

im2 = ax2.imshow(np.median(mid, axis=0), cmap='gray',vmin=500, vmax=600)
ax2.set_title('M3G20090121T025911, 158.09 K')
ax2.axis('off')

im3 = ax3.imshow(np.median(hot, axis=0), cmap='gray',vmin=500, vmax=600)
ax3.set_title('M3G20090616T073912, 166.33 K')
ax3.axis('off')

cax = fig.add_axes([0.92, 0.15, 0.03, 0.15])

cbar = fig.colorbar(im1, cax=cax)
cbar.set_label('DN')

plt.savefig("cold_vs_warm_darks.png", dpi=300) 
```
