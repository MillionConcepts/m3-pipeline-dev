```python
from astropy.io import fits
import numpy as np

obs_id = "m3g20090111t013904"  

og = "/home/bekah/m3-pipeline-dev/l0_l1b_l2/example_data/m3g20090111t013904_l1b_rdn.fits"
og_based = f"/home/bekah/m3-pipeline-dev/l0_l1b_l2/example_data/{obs_id}_old_new.fits"
new = f"/home/bekah/m3-pipeline-dev/l0_l1b_l2/example_data/{obs_id}_l1b_new_new.fits"

def diff_and_save(file1, file2, output_path):
    """Subtract file2's data from file1's data and save as a new FITS file."""
    with fits.open(file1) as hdul1, fits.open(file2) as hdul2:
        data1 = hdul1[0].data.astype(np.float64)
        data2 = hdul2[0].data.astype(np.float64)

        if data1.shape != data2.shape:
            raise ValueError(
                f"Shape mismatch: {file1} is {data1.shape}, {file2} is {data2.shape}"
            )

        diff = data1 / data2

        new_header = hdul1[0].header.copy()
        new_hdu = fits.PrimaryHDU(data=diff, header=new_header)
        new_hdu.writeto(output_path, overwrite=True)

    print(f"Saved difference ({file1} - {file2}) to {output_path}")

diff_and_save(og, og_based, "/home/bekah/m3-pipeline-dev/l0_l1b_l2/example_data/div_og_ogbased.fits")

# diff_and_save(og, new, "/home/bekah/m3-pipeline-dev/l0_l1b_l2/example_data/diff_og_new.fits")
```
